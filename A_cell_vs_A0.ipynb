```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# --- 1. Parametri ---
R_tot = 1  # Valore di saturazione (imposta a 1 per dati normalizzati)
tempi = [1, 24, 96]  # Tempi espressi in ore

# Range per A_0 (es. da 0.0001 a 0.15 h^-1)
# Puoi modificare l'intervallo in base alla scala dei tuoi dati
A_0 = np.linspace(0.0001, 0.15, 500)

# --- 2. Creazione del grafico ---
plt.figure(figsize=(10, 6))

for t in tempi:
    # Formula: R_tot * (1 - e^(-A_0 * t))
    y = R_tot * (1 - np.exp(-A_0 * t))
    plt.plot(A_0, y, label=f"t = {t} h", linewidth=2)

# --- 3. Personalizzazione e formattazione ---
plt.title(
    r"Andamento di $R_{tot} \cdot (1 - e^{-A_0 \cdot t})$ in funzione di $A_0$",
    fontsize=14,
)
plt.xlabel(r"$A_0$ ($\text{h}^{-1}$)", fontsize=12)
plt.ylabel(r"$R(t)$", fontsize=12)
plt.grid(True, linestyle="--", alpha=0.6)
plt.legend(fontsize=11)
plt.tight_layout()

# Mostra il grafico nel notebook
plt.show()
```
